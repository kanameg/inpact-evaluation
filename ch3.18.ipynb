```python
import pandas as pd
import statsmodels.formula.api as smf
from linearmodels.iv import IV2SLS
from statsmodels.iolib.summary2 import summary_col
```

```python
URL_LENTA_DATA = "https://raw.githubusercontent.com/HirotakeIto/intro_to_impact_evaluation_with_python/main/data/lenta_dataset.csv"
```

```python
abtest_df = pd.read_csv(URL_LENTA_DATA)
abtest_df.head()
```

```python
# ------------------------------------------------------------------------
# サブサンプル分割による異質性の分析
# ------------------------------------------------------------------------
```

```python
men_df = abtest_df.loc[abtest_df["is_women"] == 0, :]
women_df = abtest_df.loc[abtest_df["is_women"] == 1, :]
```

```python
result_men = smf.ols(formula="response_att ~ is_treatment", data=men_df).fit()
result_women = smf.ols(formula="response_att ~ is_treatment", data=women_df).fit()
# 推定結果をまとめて表示
summary_col([result_men, result_women], model_names=["男性", "女性"], stars=False)
```
