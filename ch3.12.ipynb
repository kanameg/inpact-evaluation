```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
URL_STRATIFIED_TRIAL = "https://raw.githubusercontent.com/HirotakeIto/intro_to_impact_evaluation_with_python/main/data/ch3_stratified_trial.csv"
```

```python
stratified_df = pd.read_csv(URL_STRATIFIED_TRIAL)
stratified_df.head()
```

```python
# ------------------------------------------------------------------------
# 階層化A/Bテストの分析（通常のA/Bテストと同様に行った場合）
# ------------------------------------------------------------------------
```

```python
result = smf.ols(formula="y ~ is_treatment", data=stratified_df).fit()
print(result.summary().tables[1])
```

```python
# ------------------------------------------------------------------------
# 階層化A/Bテストの分析（ダミー変数を利用した場合）
# ------------------------------------------------------------------------
```

```python
result = smf.ols(formula="y ~ is_treatment + C(group_name)", data=stratified_df).fit()
print(result.summary().tables[1])
```
