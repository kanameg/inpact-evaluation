```python
import numpy as np
import seaborn as sns
```

```python
def assign_strata(group_df, ratio_treat, rng, column_name):
    num_samples = round(len(group_df) * ratio_treat)
    group_df[column_name] = 0
    treat_indices = rng.choice(group_df.index, size=num_samples, replace=False)
    group_df.loc[treat_indices, column_name] = 1
    return group_df
```

```python
# データの読み込み
titanic_df = sns.load_dataset("titanic")[["survived", "sex", "pclass"]]
```

```python
# groupbyメソッドを使って無作為抽出を実行
rng = np.random.default_rng(11)
titanic_assigned_df = titanic_df.groupby(["sex", "pclass"], group_keys=False).apply(
    assign_strata, ratio_treat=0.3, column_name="is_treat", rng=rng
)
```
