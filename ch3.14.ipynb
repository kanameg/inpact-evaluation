```python
import pandas as pd
import statsmodels.formula.api as smf
from linearmodels.iv import IV2SLS
```

```python
URL_NONCOMPLIANCE = "https://raw.githubusercontent.com/HirotakeIto/intro_to_impact_evaluation_with_python/main/data/ch3_noncompliance_abtest.csv"
```

```python
noncompliance_df = pd.read_csv(URL_NONCOMPLIANCE)
noncompliance_df.head()
```

```python
# ------------------------------------------------------------------------
# 施策意図の効果の分析
# ------------------------------------------------------------------------
```

```python
# Intent to Treat (ITT) 効果の推定
result = smf.ols(formula="purchase ~ assignment", data=noncompliance_df).fit()
print(result.summary().tables[1])
```

```python
# ------------------------------------------------------------------------
# 平均的な開封割合の確認
# ------------------------------------------------------------------------
```

```python
noncompliance_df.groupby("assignment")["is_deliver"].mean()
```

```python
# ------------------------------------------------------------------------
# 試作効果の復元
# ------------------------------------------------------------------------
```

```python
iv = IV2SLS.from_formula(
    formula="purchase ~ 1 + [is_deliver ~ assignment] + x", data=noncompliance_df
).fit()
print(iv.summary.tables[1])
```
