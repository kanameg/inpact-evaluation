```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
URL_CLUSTER_TRIAL = "https://raw.githubusercontent.com/HirotakeIto/intro_to_impact_evaluation_with_python/main/data/ch3_cluster_trial.csv"
```

```python
cluster_trial_df = pd.read_csv(URL_CLUSTER_TRIAL)
cluster_trial_df.head()
```

```python
# ------------------------------------------------------------------------
# 割当単位と分析単位が不一致なシミュレーションデータでの回帰分析
# ------------------------------------------------------------------------
```

```python
result = smf.ols(formula="is_click ~ is_treatment", data=cluster_trial_df).fit()
print(result.summary().tables[1])
```
