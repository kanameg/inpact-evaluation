```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
URL_AATEST = "https://raw.githubusercontent.com/HirotakeIto/intro_to_impact_evaluation_with_python/main/data/ch3_aatest_trial.csv"
```

```python
aa_test_df = pd.read_csv(URL_AATEST)
aa_test_df.head()
```

```python
# ------------------------------------------------------------------------
# A/Aテストの結果を回帰分析で確認する（成功例）
# ------------------------------------------------------------------------
```

```python
result = smf.ols(formula="is_click ~ is_treatment", data=aa_test_df).fit()
print(result.summary().tables[1])
```
